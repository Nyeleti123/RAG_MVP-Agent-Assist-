```python
import faiss
import numpy as np
import pandas as pd

from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer(
    "all-mpnet-base-v2"
)
```

```python
index = faiss.read_index(
    "../vector_db/faiss.index"
)

metadata = pd.read_parquet(
    "../vector_db/metadata.parquet"
)

print(index.ntotal)
```

```python
query = "How does the call centre handle refunds?"

query_embedding = model.encode(

    [query],

    convert_to_numpy=True,

    normalize_embeddings=True

)
```

```python
k = 5

scores, indices = index.search(
    query_embedding,
    k
)
```

```python
print(indices)
```

```python
results = metadata.iloc[
    indices[0]
]

results
```

```python
for rank, (_, row) in enumerate(results.iterrows(), start=1):

    print("=" * 80)

    print(f"Result {rank}")

    print(f"Document : {row['filename']}")

    print(f"Chunk : {row['chunk_index']}")

    print()

    print(row["chunk_text"][:800])

    print()
```
